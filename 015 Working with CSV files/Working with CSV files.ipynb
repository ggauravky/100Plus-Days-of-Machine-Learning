```python
# set parameter
# df = pd.read_csv('cheating_network_edges.tsv', sep='\t')
df=pd.read_csv('cheating_network_edges.csv',names=['source', 'target'] )
df
```

```python
# index_col parameter
df=pd.read_csv('cheating_network_edges.csv', index_col=0)
df
```

```python
# header parameter
df=pd.read_csv('cheating_network_edges.csv', header=1)
df
```

```python
# use_cols parameter
df=pd.read_csv('cheating_network_edges.csv', usecols=['source', 'target'])
df
```

```python
# skiprows parameter
df=pd.read_csv('cheating_network_edges.csv', skiprows=[0,1])
df
# # nrows parameter
df=pd.read_csv('cheating_network_edges.csv', nrows=5)
df
```

```python
# encoding parameter
df=pd.read_csv('cheating_network_edges.csv', encoding='utf-8')
df
```

```python
# skip bad lines parameter
df=pd.read_csv('cheating_network_edges.csv', error_bad_lines=False)
df
```

```python
# dtypes parameter
df=pd.read_csv('cheating_network_edges.csv', dtype={'source': str, 'target': str})
df
```

```python
# handling dates
df=pd.read_csv('cheating_network_edges.csv', parse_dates=['date'])
df
```

```python
# convertors parameter
df=pd.read_csv('cheating_network_edges.csv', converters={'source': str, 'target': str})
df
```

```python
# na_values parameter
df=pd.read_csv('cheating_network_edges.csv', na_values=['NA', 'NaN'])
df
```

```python
# load huge files in chunks
chunk_size = 1000
for chunk in pd.read_csv('cheating_network_edges.csv', chunksize=chunk_size):
    # Process each chunk here
    print(chunk)
```
