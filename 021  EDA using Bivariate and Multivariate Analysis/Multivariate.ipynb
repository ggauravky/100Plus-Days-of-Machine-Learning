```python
import pandas as pd
import seaborn as sns
import matplotlib.pyplot as plt
```

```python
titanic = pd.read_csv('titanic.csv')
flights = pd.read_csv('flights.csv')
iris = pd.read_csv('iris.csv')
tips = pd.read_csv('tip.csv')
```

```python
tips.head()
```

```python
sns.scatterplot(
    data=tips,
    x='total_bill',
    y='tip',
    hue='sex',
    style='smoker',
    size='size'
)
```

```python
# bar  plot

titanic.head()
```

```python
sns.barplot(
    data=titanic,
    x='Pclass',
    y='Fare',
    hue='Sex'
)
```

```python
# blox plot
sns.boxplot(
    data=titanic,
    x='Sex',
    y='Age',
    hue='Survived'
)
```

```python
# Distplot
sns.displot(
    data=titanic,
    x='Age',
    hue='Sex',
    kind='kde',
)
```

```python
sns.heatmap(
    pd.crosstab(titanic['Pclass'], titanic['Survived']),
    annot=True,
    fmt='d'
)
```

```python
survival_rate = titanic.groupby('Pclass').mean(numeric_only=True)['Survived'] * 100

survival_rate
```

```python
survival_rate = titanic.groupby('Pclass')['Survived'].mean() * 100

survival_rate.plot(kind='bar')
plt.ylabel('Survival Rate (%)')
plt.xlabel('Passenger Class')
plt.title('Survival Rate by Passenger Class')
plt.show()
```
